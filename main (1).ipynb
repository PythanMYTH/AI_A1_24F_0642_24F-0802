```python

## Controls

Left Mouse Click:
- First click sets the Start node
- Second click sets the Target node
- Subsequent clicks create walls

Right Mouse Click:
- Reset a cell

Keyboard Controls:

B  - Run Breadth First Search  
D  - Run Depth First Search  
U  - Run Uniform Cost Search  
L  - Run Depth Limited Search  
I  - Run Iterative Deepening DFS  
N  - Run Bidirectional Search  
C  - Clear the grid  

## Visualization Details

Each node in the grid is represented using a specific color:

Start node  
Target node  
Wall  
Frontier node  
Explored node  
Final path  

The GUI updates after every expansion step to show the exploration pattern of each algorithm.

## Test Cases

Each algorithm was tested in two scenarios:

Best case – where the target is close to the start.  
Worst case – where the algorithm explores most of the grid.

Screenshots of these test cases are included in the report.

## Author
Name:  
Student ID:




import pygame
from grid import make_grid
from gui import draw
from algorithms.bfs import bfs
from algorithms.dfs import dfs
from algorithms.ucs import ucs
from algorithms.dls import dls
from algorithms.iddfs import iddfs
from algorithms.bidirectional import bidirectional

WIDTH = 800
ROWS = 40

def main():
    pygame.init()
    WIN = pygame.display.set_mode((WIDTH, WIDTH))
    pygame.display.set_caption("GOOD PERFORMANCE TIME APP")

    grid = make_grid(ROWS, WIDTH)

    start = None
    target = None
    run = True

    while run:
        draw(WIN, grid, ROWS, WIDTH)

        for event in pygame.event.get():

            if event.type == pygame.QUIT:
                run = False

            # 🖱️ MOUSE CONTROLS
            if pygame.mouse.get_pressed()[0]:
                pos = pygame.mouse.get_pos()
                row = pos[1] // (WIDTH // ROWS)
                col = pos[0] // (WIDTH // ROWS)
                cell = grid[row][col]

                if not start and cell != target:
                    start = cell
                    start.make_start()

                elif not target and cell != start:
                    target = cell
                    target.make_target()

                elif cell != start and cell != target:
                    cell.make_wall()

            # RIGHT CLICK → RESET CELL
            elif pygame.mouse.get_pressed()[2]:
                pos = pygame.mouse.get_pos()
                row = pos[1] // (WIDTH // ROWS)
                col = pos[0] // (WIDTH // ROWS)
                cell = grid[row][col]
                cell.reset()

                if cell == start:
                    start = None
                elif cell == target:
                    target = None

            # ⌨️ KEY CONTROLS
            if event.type == pygame.KEYDOWN:

                if event.key == pygame.K_b and start and target:
                    bfs(lambda: draw(WIN, grid, ROWS, WIDTH), grid, start, target)

                if event.key == pygame.K_d and start and target:
                    dfs(lambda: draw(WIN, grid, ROWS, WIDTH), grid, start, target)

                if event.key == pygame.K_u and start and target:
                    ucs(lambda: draw(WIN, grid, ROWS, WIDTH), grid, start, target)

                if event.key == pygame.K_l and start and target:
                    dls(lambda: draw(WIN, grid, ROWS, WIDTH), grid, start, target, limit=20)

                if event.key == pygame.K_i and start and target:
                    iddfs(lambda: draw(WIN, grid, ROWS, WIDTH), grid, start, target, max_depth=40)

                if event.key == pygame.K_n and start and target:
                    bidirectional(lambda: draw(WIN, grid, ROWS, WIDTH), grid, start, target)

                # CLEAR GRID
                if event.key == pygame.K_c:
                    start = None
                    target = None
                    grid = make_grid(ROWS, WIDTH)

    pygame.quit()

if __name__ == "__main__":
    main()
```
