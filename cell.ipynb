```python
class Cell:
    def __init__(self, row, col, size):
        self.row = row
        self.col = col
        self.x = col * size
        self.y = row * size
        self.color = WHITE
        self.neighbors = []

    def is_wall(self):
        return self.color == BLACK

    def make_start(self): self.color = ORANGE
    def make_target(self): self.color = TURQUOISE
    def make_wall(self): self.color = BLACK
    def make_frontier(self): self.color = PURPLE
    def make_explored(self): self.color = BLUE
    def make_path(self): self.color = YELLOW

    def __hash__(self):
        return hash((self.row, self.col))

    def __eq__(self, other):
        return self.row == other.row and self.col == other.col
```
