```python
def dfs(draw, grid, start, target):

    stack = [start]
    visited = {start}
    parent = {}

    while stack:

        current = stack.pop()

        #  If target found
        if current == target:
            reconstruct_path(parent, target, draw)
            return True

        # Explore neighbors (REVERSED for correct DFS order)
        for neighbor in reversed(get_neighbors(current, grid)):

            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current
                stack.append(neighbor)

                neighbor.make_frontier()

        if current != start:
            current.make_explored()

        draw()   # animate step-by-step

    return False


def get_neighbors(cell, grid):
    rows = len(grid)
    cols = len(grid[0])
    r, c = cell.row, cell.col

    directions = [
        (-1, 0),   # Up
        (0, 1),    # Right
        (1, 0),    # Down
        (1, 1),    # Bottom-Right
        (0, -1),   # Left
        (-1, -1),  # Top-Left
        (-1, 1),   # Top-Right
        (1, -1)    # Bottom-Left
    ]

    neighbors = []

    for dr, dc in directions:
        nr, nc = r + dr, c + dc

        if 0 <= nr < rows and 0 <= nc < cols:
            if not grid[nr][nc].is_wall():
                neighbors.append(grid[nr][nc])

    return neighbors
```
