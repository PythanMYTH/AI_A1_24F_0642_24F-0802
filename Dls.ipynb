```python
from utils.helpers import get_neighbors, reconstruct_path

def dls(draw, grid, start, target, limit):

    stack = [(start, 0)]
    parent = {}
    visited = {start}

    while stack:

        current, depth = stack.pop()

        if current == target:
            reconstruct_path(parent, target, draw)
            return True

        if depth < limit:

            for neighbor in reversed(get_neighbors(current, grid)):

                if neighbor not in visited:
                    visited.add(neighbor)
                    parent[neighbor] = current
                    stack.append((neighbor, depth + 1))

                    neighbor.make_frontier()

        if current != start:
            current.make_explored()

        draw()

    return False
```
