```python
def bfs(draw, grid, start, target):

    queue = deque([start])
    visited = {start}
    parent = {}

    while queue:

        current = queue.popleft()

      # if target is found
        if current == target:
            reconstruct_path(parent, target, draw)
            return True

        for neighbor in get_neighbors(current, grid):

            if neighbor not in visited:

                parent[neighbor] = current
                visited.add(neighbor)
                queue.append(neighbor)

                neighbor.make_frontier()

        # avoid recoloring start node
        if current != start:
            current.make_explored()

        draw()   # animate step-by-step

    return False
```
