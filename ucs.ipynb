```python
import heapq
from utils.helpers import get_neighbors, reconstruct_path

def ucs(draw, grid, start, target):

    pq = []
    heapq.heappush(pq, (0, start))

    cost = {start: 0}
    parent = {}

    while pq:

        current_cost, current = heapq.heappop(pq)

        if current == target:
            reconstruct_path(parent, target, draw)
            return True

        for neighbor in get_neighbors(current, grid):

            new_cost = current_cost + 1   # change to 1.4 for diagonal if required

            if neighbor not in cost or new_cost < cost[neighbor]:

                cost[neighbor] = new_cost
                parent[neighbor] = current

                heapq.heappush(pq, (new_cost, neighbor))
                neighbor.make_frontier()

        if current != start:
            current.make_explored()

        draw()

    return False
```
