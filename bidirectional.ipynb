```python
from collections import deque
from utils.helpers import get_neighbors, reconstruct_path

def bidirectional(draw, grid, start, target):

    q_start = deque([start])
    q_target = deque([target])

    parent_start = {}
    parent_target = {}

    visited_start = {start}
    visited_target = {target}

    while q_start and q_target:

        current_start = q_start.popleft()
        current_target = q_target.popleft()

        if current_start in visited_target:
            reconstruct_path(parent_start, current_start, draw)
            reconstruct_path(parent_target, current_start, draw)
            return True

        if current_target in visited_start:
            reconstruct_path(parent_start, current_target, draw)
            reconstruct_path(parent_target, current_target, draw)
            return True

        for neighbor in get_neighbors(current_start, grid):

            if neighbor not in visited_start:
                visited_start.add(neighbor)
                parent_start[neighbor] = current_start
                q_start.append(neighbor)
                neighbor.make_frontier()

        for neighbor in get_neighbors(current_target, grid):

            if neighbor not in visited_target:
                visited_target.add(neighbor)
                parent_target[neighbor] = current_target
                q_target.append(neighbor)

        draw()

    return False
```
